```python
import pandas as pd
import pickle

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df=pd.read_csv('clean_resume.csv')
```

```python
df.head()
```

```python
print(df.columns)
```

```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'[^a-zA-Z ]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text

df["Clean_Resume"] = df["Resume"].apply(clean_text)
```

```python
tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(df["Clean_Resume"])
```

```python
print(df.columns)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf.fit_transform(df["Clean_Resume"])
```

```python
pickle.dump(tfidf, open("tfidf.pkl", "wb"))
pickle.dump(df, open("resume_data.pkl", "wb"))

print("Model Saved Successfully!")
```

```python
def recommend_jobs(resume_text):

    # Convert input resume into TF-IDF vector
    resume_vector = tfidf.transform([resume_text])

    # Calculate similarity with all resumes
    scores = cosine_similarity(resume_vector, tfidf_matrix)

    # Get top 5 similar resumes
    top_index = scores.argsort()[0][-5:][::-1]

    # Return recommended categories
    return df.iloc[top_index][["Category"]]
```

```python
sample_resume = """
Python
Machine Learning
SQL
Data Analysis
"""

recommend_jobs(sample_resume)
```
